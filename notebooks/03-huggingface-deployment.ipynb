```python
MODEL_PATH = "/kaggle/input/datasets/irakozegaddiel/fineduqa-flan-t5-final-model/flan_t5_exp2_final"
OOD_ARTIFACT_PATH = "/kaggle/input/notebooks/irakozegaddiel/02-flan-t5-modeling/ood_detector_artifacts.pt"
OOD_CONFIG_PATH = "/kaggle/input/notebooks/irakozegaddiel/02-flan-t5-modeling/ood_detector_config.json"
```

```python
import os

print("Model:", os.path.exists(MODEL_PATH))
print("OOD artifacts:", os.path.exists(OOD_ARTIFACT_PATH))
print("OOD config:", os.path.exists(OOD_CONFIG_PATH))
```

### Upload FLAN-T5 to Hugging Face

```python
HF_USERNAME = "Gaddiel2005"
MODEL_REPO = f"{HF_USERNAME}/FinEduQA"
```

```python
from huggingface_hub import HfApi

api = HfApi(token=hf_token)

api.upload_folder(
    folder_path=MODEL_PATH,
    repo_id=MODEL_REPO,
    repo_type="model",
    commit_message="Upload final fine-tuned FLAN-T5 model"
)

print("Model uploaded successfully.")
```

### Upload the OOD detector

```python
api.upload_file(
    path_or_fileobj=OOD_ARTIFACT_PATH,
    path_in_repo="ood/ood_detector_artifacts.pt",
    repo_id=MODEL_REPO,
    repo_type="model",
    commit_message="Add OOD detector embeddings"
)

api.upload_file(
    path_or_fileobj=OOD_CONFIG_PATH,
    path_in_repo="ood/ood_detector_config.json",
    repo_id=MODEL_REPO,
    repo_type="model",
    commit_message="Add OOD detector configuration"
)

print("OOD detector uploaded successfully.")
```
